```python
### LAB 1 ###
from astropy.io import fits
import matplotlib.pyplot as plt
import glob
import numpy as np
```

```python
## FUNCTIONS ##

# Function that corrects for 12 bit measurements of CMOS sensor #
def real_cts(m):
    return (m/16) + 1

# Image Average func. #
def img_average(array):
    mean = np.mean(array,axis=(0))
    return mean
```

```python
## FUNCTION TO NORMALIZE FLATS ##
def normiept1(data, bias):
    denom = []
    med_bias = np.median(bias)
    for i in range(len(data)):
        for j in range(4656):
            #calculates data at pixel minus average bias
            denom.append((data[i,j] - med_bias))
    #print(denom)
    return denom

def normiept2(data, bias, denom):
    med_denom = np.median(denom)
    z = (np.zeros([len(data),4656]))
    for p in range(len(data)):
        for q in range(4656):
            newz = (data[p,q] - bias[p,q])/(med_denom) #normalization equation
            z[p,q] += newz
    return z
```

## --Random Error--

```python
bias1a = fits.open('OfMeatballs/bias-0005.fit')
bias1 = real_cts(bias1a[0].data)
bias2a = fits.open('OfMeatballs/bias-0019.fit')
bias2 = real_cts(bias2a[0].data)
```

```python
def ranerr(bias1, bias2):
    dif_img = np.zeros([len(bias1), len(bias1)])
    for i in range(len(bias1)):
        for j in range(len(bias1)):
            diff = bias1[i,j] - bias2[i,j]
            dif_img[i,j] += diff
    return dif_img      
```

```python
chicken = ranerr(bias1, bias2)
```

```python
## PLOT OF random ERROR ##
plt.imshow(chicken[100:200,100:200], cmap = 'viridis', extent= [100,200,100,200])
plt.colorbar(label = 'Counts')
plt.xlabel('X-Position on Camera (pixels)')
plt.ylabel('Y-Position on Camera (pixels)')
plt.title('Read Noise Image: QHY 163M CMOS')
plt.savefig('read_noise_img.png')
```

```python
## Function to calculate read noise ##
def readnoise(diff):
    sigmaread = np.zeros([4656, 3522])
    sqrt2 = float(np.sqrt(2))
    for i in range(len(diff)):
        for j in range(len(diff)):
            #newsig = diff[i,j]/sqrt2
            sigmaread[i,j] += (diff[i,j]/sqrt2)
    return sigmaread
```

```python
## Histogram plot of difference image ##
egg = readnoise(chicken)
plt.hist(egg)
plt.xlabel('Count Value')
plt.ylabel('# of Counts')
plt.title('Read Noise Histogram: QHY 163M CMOS')
plt.savefig('read_noise_hist.png')
```

## --Correcting Science Images--

```python
## Function that corrects science images ##
def sciimgfunc(data, avgdark, avgbias, normflat):
    sci_correct = np.zeros([3522,4656])
    for i in range(3522):
        for j in range(4656):
            sci_correct[i,j] = ((data[i,j]- avgdark[i,j])-avgbias[i,j])/(normflat[i,j])
    return sci_correct
```
